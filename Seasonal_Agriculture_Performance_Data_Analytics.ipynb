## 6. Missing Values & Duplicates

```python
# misssing values
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print("Columns with missing values:")
print(missing)
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
print(f"Percentage of total cells: {(df.isnull().sum().sum() / df.size) * 100:.2f}%")
```

```python
for col in ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']:
    df[col] = df.groupby('Season')[col].transform(lambda x: x.fillna(x.median()))

print("Missing values after imputation:")
print(df.isnull().sum().sum())
```

```python
# checking for duplicates
duplicates = df.duplicated().sum()
print("Number of duplicate rows:", duplicates)

df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

## 7. Outlier detection

```python
# Boxplots for key numeric columns
cols_to_check = ['Yield_Tonnes_Ha', 'Total_Cost_INR', 'Revenue_INR', 'Profit_INR', 'Rainfall_mm']

fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for i, col in enumerate(cols_to_check):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)
plt.tight_layout()
plt.show()
```

```python
# Count outliers using IQR method
for col in cols_to_check:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(col, "-> outliers:", len(outliers))
```

### Outlier Decision
Outliers were detected in Yield_Tonnes_Ha, Revenue_INR and Profit_INR using the IQR method.
On inspection, these are not data errors - they reflect real variation across different crops
(e.g. Sugarcane vs Chilli yield scales) and genuine loss-making records (negative profit).
These outliers are kept in the dataset since removing them would hide real patterns relevant
to the seasonal analysis.

## 8. Feature / Variable Review

The dataset's 28 columns can be grouped into five categories based on what they represent.

**Identifier variables:** Farm_ID, State, District
These identify each record and its location. They don't directly affect performance but
are useful for grouping and regional comparisons.

**Grouping / categorical variables:** Crop, Season, Irrigation_Method
These describe *what* was grown and *under what conditions*. Season is the most important
variable in this project, since the entire analysis is built around comparing performance
across Kharif, Rabi and Zaid.

**Environmental variables:** Farm_Area_Hectares, Rainfall_mm, Avg_Temperature_C,
Humidity_pct, Sunlight_Hours_Day, Soil_pH, Soil_Moisture_pct
These are the natural/farm conditions that change from season to season and directly affect
crop growth. They help explain *why* performance differs across seasons, not just *that* it
differs.

**Input / farming practice variables:** Nitrogen_kg_ha, Phosphorus_kg_ha, Potassium_kg_ha,
Fertilizer_kg_ha, Pesticide_Litre_ha, Seed_Quality_Score, Water_Used_m3,
Water_Efficiency_t_per_1000m3
These show how much effort and resources farmers put in. Comparing these across seasons
shows whether farming practices change along with the season, and whether more input
actually leads to better output.

**Outcome / performance variables:** Yield_Tonnes_Ha, Production_Tonnes,
Market_Price_INR_Tonne, Total_Cost_INR, Revenue_INR, Profit_INR, Disease_Pest_Risk_pct
These are the actual results of farming and the main focus of the analysis. Yield and
Profit are the two most direct measures of "performance" as asked in the problem statement,
so most of the seasonal comparisons will center on these two.

**Most relevant for this problem:** Season (the grouping variable), Yield_Tonnes_Ha and
Profit_INR (the performance outcomes), and Rainfall_mm / Avg_Temperature_C /
Soil_Moisture_pct (the environmental drivers most likely to explain seasonal differences).

## 13. Additional Student-Driven Analysis

Three additional analyses were conducted, each investigating a different angle of the
seasonal performance problem: regional consistency, disease/pest risk, and water use
efficiency.

### Analysis 1



```python
# Analysis 1
# Question: Is Zaid's low profitability seen across all states, or only in some?
# Analysis: Group profit by State and Season to compare regional patterns.

state_season_profit = df.groupby(['State', 'Season'])['Profit_INR'].mean().unstack()
print(state_season_profit)

plt.figure(figsize=(8, 5))
sns.heatmap(state_season_profit, annot=True, fmt='.0f', cmap='RdYlGn', center=0)
plt.title('Average Profit by State and Season')
plt.show()
```

 Interpretation: Kharif is profitable in every state without exception. Zaid isunprofitable in 5 of 8 states (worst in Gujarat and Andhra Pradesh), but stays positive in Karnataka, Punjab, and marginally in Madhya Pradesh. So Zaid's weakness is widespread but not universal - it is a regional risk, not a fixed rule.

### Analysis 2

```python
# Analysis 2
# Question: Does higher disease/pest risk explain why some seasons earn less profit?
# Analysis: Compare average Disease_Pest_Risk_pct across seasons and plot it against Profit.

risk_by_season = df.groupby('Season')['Disease_Pest_Risk_pct'].mean()
print(risk_by_season)

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Disease_Pest_Risk_pct', y='Profit_INR', hue='Season', alpha=0.5)
plt.title('Disease/Pest Risk vs Profit by Season')
plt.axhline(0, color='black', linewidth=0.8)
plt.show()
```

 Interpretation: Kharif actually has the highest average pest risk (54.47%) of all three seasons, yet remains the most profitable. This rules out pest/disease risk as the cause of Zaid's weaker performance - the relationship runs the opposite way.

### Analysis 3

```python
# Student Analysis 3
# Question: Is Zaid's weaker performance linked to how efficiently water is used?
# Analysis: Compare average water used and water efficiency across seasons.

water_eff = df.groupby('Season')[['Water_Used_m3', 'Water_Efficiency_t_per_1000m3']].mean()
print(water_eff)

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Season', y='Water_Efficiency_t_per_1000m3')
plt.title('Water Efficiency by Season')
plt.show()
```

Interpretation: Zaid uses the most water on average (6,419.89 m3) but has the lowest water efficiency (4.41 t per 1000m3), compared to Kharif's 5.89. Zaid farms spend more water for less output, making irrigation planning a concrete lever to improve outcomes.

## 15. Recommendations

1. **Prioritize Drip irrigation in Zaid season.**
   Finding: Drip-irrigated farms stayed profitable in Zaid while Flood- and Rainfed-irrigated
   farms showed negative average profit in that season (Insight 4).
   This recommendation is limited to irrigation choice, it does not assume every farm can
   switch method, since cost of conversion and local water access were not part of this dataset.

2. **Target water-efficiency improvements at Zaid-season farms.**
   Finding: Zaid used the most water on average but had the lowest water efficiency of the
   three seasons (Insight 5).
   This does not assume a specific cause of the inefficiency - only that Zaid is the season
   where efficiency gains would have the most room to help.

3. **Give closer attention to Zaid-season farming in Gujarat and Andhra Pradesh.**
   Finding: These two states showed the steepest average losses in Zaid season
   (Insight 3).
   This is a data-supported priority, not a claim that other states need no attention -
   Zaid was unprofitable in 5 of 8 states overall.

4. **Consider Chilli and Sugarcane where crop choice is flexible.**
   Finding: Both crops showed consistently higher average profit across all three seasons
   compared to Wheat, Maize, Pulses and Rice (Insight 7).
   This does not account for land suitability, market demand, or farmer preference, which
   were outside the scope of this dataset - so it is a data-supported option, not a
   universal directive.

5. **Do not prioritize pest/disease control as the primary fix for Zaid's weaker performance.**
   Finding: Kharif had the highest average pest risk of all seasons yet remained the most
   profitable, indicating pest risk does not explain the seasonal profit gap (Insight 6).
   Pest management still matters for other reasons - this only says it is not the leading
   explanation for Zaid's underperformance based on this data.

## 16. Conclusion

 Agricultural performance varies meaningfully across the
three seasons in this dataset. Kharif consistently shows the strongest performance, Rabi
is moderate, and Zaid consistently underperforms, being the only season with a negative
average profit.

 Zaid's weaker performance is not explained by pest/disease
risk (Kharif has higher risk but higher profit), but is closely associated with irrigation
method and water-use efficiency - Drip-irrigated farms remain profitable in Zaid while
Flood- and Rainfed-irrigated farms do not.

 The irrigation-by-season profit breakdown and the
water-efficiency comparison were the most actionable findings, since they point to a
specific, changeable factor (irrigation method) rather than an unchangeable one (the season
itself).

 Zaid has a notably smaller sample size (594 records) than
Kharif or Rabi, the dataset covers only one time period with no year-over-year data, and
all relationships identified are associative, not causal - irrigation method and water
efficiency are linked to profit but were not tested as direct causes of it.

 These findings could guide a focused follow-up
study on irrigation conversion costs and feasibility for Zaid-season farms, particularly
in Gujarat and Andhra Pradesh, and could support multi-year data collection to confirm
whether these seasonal patterns are consistent over time.
